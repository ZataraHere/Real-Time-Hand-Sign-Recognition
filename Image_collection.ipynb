__Importing Libraries__

```python
import cv2

import uuid

import os

import time
```

__Defining Objects__

```python
labels = ['Face', 'ThumbsUp', 'iLoveYou', 'Hello', 'ThankYou', 'Yes', 'No', 'Victory']
```

__Setting Up Folders__

```python

IMAGES_PATH= os.path.join('Tensorflow','Workspace','Images','Collected_Images')
```

```python
if not os.path.exists(IMAGES_PATH):
    os.makedirs(IMAGES_PATH)
    
    for label in labels:
        location = os.path.join(IMAGES_PATH,label)

        if not os.path.exists(location):
            os.makedirs(location)
```

__Capturing Data__

```python
total_images = 2

for label in labels:

    cap = cv2.VideoCapture(0)

    print("Collecting images for {}".format(label))
    time.sleep(2)

    for image_num in range(total_images):
        ret, frame = cap.read()
        print('Collecting images {}'.format(image_num))

        
        image_name=os.path.join(IMAGES_PATH, label, f"{label}_{image_num}_{uuid.uuid1()}.jpg")
        if not os.path.exists(image_name):
            os.makedirs(image_name)
        cv2.imwrite(image_name, frame)
        time.sleep(1.5)                   

        if cv2.waitKey(1) and 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```

```python

total_images = 20  # Number of images to capture per label

for label in labels:
    cap = cv2.VideoCapture(0)  # Open the webcam

    print("Collecting images for {}".format(label))
    time.sleep(2)

    # Ensure the directory exists before saving images
    label_dir = os.path.join(IMAGES_PATH, label)
    os.makedirs(label_dir, exist_ok=True)  

    for image_num in range(total_images):
        ret, frame = cap.read()  # Capture frame
        if not ret:
            print("Failed to capture image")
            continue  # Skip saving if frame is not valid

        print('Collecting image {}'.format(image_num))

        image_name=os.path.join(IMAGES_PATH,label,label+'.'+'{}.jpg'.format(str(uuid.uuid1())))
        cv2.imwrite(image_name, frame) 

        time.sleep(1.5)  

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()

cv2.destroyAllWindows()
```
